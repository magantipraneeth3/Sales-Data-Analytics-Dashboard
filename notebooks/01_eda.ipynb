# Sales Analytics — Exploratory Data Analysis
This notebook profiles the 1,000-row sales dataset, checks data quality and explores revenue, units, regions, categories and channels.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
df = pd.read_csv(ROOT / 'data' / 'raw' / 'sales_data.csv', parse_dates=['Sale_Date'])
df.head()
```

```python
df.info()
print('Missing values:\n', df.isna().sum())
print('Duplicate rows:', df.duplicated().sum())
```

```python
df['Revenue'] = df['Quantity_Sold'] * df['Unit_Price']
df['Cost'] = df['Quantity_Sold'] * df['Unit_Cost']
df['Profit'] = df['Revenue'] - df['Cost']
df['Margin'] = np.where(df['Revenue'] != 0, df['Profit']/df['Revenue']*100, 0)

print(f"Revenue: ₹{df.Revenue.sum():,.2f}")
print(f"Profit: ₹{df.Profit.sum():,.2f}")
print(f"Orders: {len(df):,}")
print(f"Units: {df.Quantity_Sold.sum():,}")
```

```python
monthly = df.groupby(df.Sale_Date.dt.to_period('M')).Revenue.sum()
monthly.plot(figsize=(12,4), marker='o', title='Monthly Revenue')
plt.ylabel('Revenue')
plt.grid(alpha=.2)
plt.show()
```

```python
region = df.groupby('Region').Revenue.sum().sort_values(ascending=False)
category = df.groupby('Product_Category').Revenue.sum().sort_values(ascending=False)
channel = df.groupby('Sales_Channel').Revenue.sum().sort_values(ascending=False)
print('Region\n', region)
print('Category\n', category)
print('Channel\n', channel)
```

## Key EDA questions
- Which regions generate the most revenue?
- Which product categories lead revenue?
- How does revenue change over time?
- Which channel and customer type contribute most?
- Are unit price and unit cost consistent enough for profitability modeling?
